```python
#Create dataset will be used
students = ["John","Mary","Peter","Jane","David","Amy","Larry","Peter","Bob","Lane"]
marks = [60, 45, 67, 35,90,88,43,75,34,83]
```

```python
#Create counters and running total
passed = 0
failed = 0
total = 0 
```

```python
#Process data
for student, mark in zip(students,marks):
    if mark >= 60:
        result = "Pass"
        passed += 1
    else: 
        result = "Fail"
        failed += 1
    total += mark
    print(student, mark, result)
```

```python
# Calculate the average and display the summary
average = total / len(marks)

print("Students passed:" , passed)
print("Students failed:" , failed)
print("Total marks:", total)
print("Class Average:", average)
```

```python
students_over_80 = 0

for student, mark in zip(students,marks):
    if mark >=80:
        students_over_80 += 1
students_over_80
```

```python
highest_mark = max(zip(marks, students))
highest_mark
```

```python
#Calculate Percentage of students who passed

passed_per = (passed / len(students)) * 100
print(f"{passed_per} % of students have passed.")
```
