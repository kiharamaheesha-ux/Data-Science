## Trial Assignment:
- Imagine you are a data analyst and you are to segment customers based on their annual spending
- ask the customers their annual spending
- then segment the customers into:
       - Over 20k = Premium
  - 10K - 20K = gOLD
  - 5K - 10K = Silver
  - Rest are standard

```python
customers = ["Mark", "Mary", "Lane", "John", "Pat"]
spendings = [2000, 40530, 10404,6040,20435]
```

```python
for customers, spendings in zip(customers,spendings):
    if spendings > 20000:
        customer_rank = "Premium"
    elif spendings >= 10000:
        
        customer_rank = "Gold"
    elif spendings >= 5000:
        
        customer_rank = "Silver"
    else:
        customer_rank = "Standard"
    print(f"$", spendings, customers, customer_rank)
```
