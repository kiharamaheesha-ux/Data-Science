##Lists
- An Ordered changeable, mutable collection of elements that allows duplicates

```python
expenses = [45.50, 12.00, 89.99 , 12.00]
```

```python
expenses
```

```python
# Add elements in a list - appending adding elements at the last 
expenses.append(30.00)
```

```python
expenses
```

```python
#removing elements from a List
expenses.remove(12.00) # Remove the first that it will encounter
```

```python
expenses
```

```python
#Sort
expenses.sort()
expenses
```

```python
#Operators like suming
sum(expenses)
```

```python
# Find the maximum
max(expenses)
```

#List comprehension


```python
double = [e * 2 for e in expenses]
double
```

```python
#walk through the list and return expenses greater than 50
large_expenses = [e for e in expenses if e > 50]
large_expenses
```

## Tuples

## Dictionaries
- Unordered collection of key-value pairs

```python
transaction = {
    "description" : "Groceries" ,
    "amount" : 45.50,
    "category" : "Food",
    "date" : "2026-09-30"
}
```

```python
transaction
```

```python
#Accessing a value with the key
transaction["amount"]
```

```python
# Adding to the dictionary
```

```python
transaction["notes"] = "Weekly shop" 
```

```python
transaction
```

```python
#Change the value --> key
transaction["amount"] = 89.95
transaction
```

```python
#looping through the dictionary
for key, value in transaction.items():
    print(f"{key} : {value}")
```

```python
# A list of transaction
transactions = [
      {"description":"groceries","amount": 45, "category": "food"},
      {"description":"bus pass","amount": 30.00, "category": "transport"},
      {"description":"rent","amount": 800, "category": "housing"}
]
```

```python
#Find the total spent for each transaction
total = sum(t["amount"] for t in transactions)
print(f"Total spent: {total}")
```
