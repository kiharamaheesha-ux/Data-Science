# Create a mini project
## A student performance analytics
- Create a python program that processes students marks, determines who passed, counts the results and then calculate the class average



```python
#Create dataset will be used
students = ["John","Mary","Peter","Jane","David"]
marks = [60, 45, 67, 35,90]
```

```python
#Create counters and running total
passed = 0
failed = 0
total = 0 
```

```python
#Process data
for student, mark in zip(students,marks):
    if mark >= 50:
        result = "Pass"
        passed += 1
    else: 
        result = "Fail"
        failed += 1
    total += mark
    print(student, mark, result)
```

```python
# Calculate the average and display the summary
average = total / len(marks)

print("Students passed:" , passed)
print("Students failed:" , failed)
print("Total marks:", total)
print("Class Average:", average)
```
