# Topic 5: Functions and Modular Programming

Functions let you write code and reuse it anywhere

Topics: 
- Definning functions
- Arguments
- Default/keyword args
- Lamba
- Scope
- Modules

```python
def greet():
    print("Hello!")
```

```python
#Calling the function
greet()
```

```python
#Parameters
def greet(name):
    print(f"Hello, {name}!")
```

```python
#Calling the fucntion with argument
greet("Kihara")
```

```python
greet("Claire")
```

```python
## a function can take multiple parameters
def introduce(name, age):
    print(f"My name is {name}.")
    print(f"I am {age} years old.")
```

```python
introduce("John",35)
```

```python
# Positional Parameters
introduce(43,"Mary")
```

```python
introduce(age = 43, name = "Polo")
```

```python
#Default parameters
def welcome(name = "User"):
    print(f"Welcome {name}!")
```

```python
welcome("Kihara")
```

```python
welcome()
```

```python
# Difference of return and print
def add_print(a,b):
    print(a+b)
```

```python
result = add_print(10,20)
print("Stored result:", result)
```

```python
# Return
def add_return(a,b):
    return a + b
```

```python
result = add_return(20,30)
print("Stored result:", result)
```
